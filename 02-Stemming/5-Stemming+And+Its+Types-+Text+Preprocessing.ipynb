```python
`## Classification problem
## Comments of product is a positive or negative review
## Reviews  ---> [eating, eat, eaten] --> eat
### Quick text normalization, search indexing
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
## Porter Stemmer
from nltk.stem import PorterStemmer
```

🌱 Stemming vs Lemmatization
Aspect	Stemming	Lemmatization
Meaning	Cuts words to their root form	Converts words to their dictionary (base) form
Example	running → run	running → run
But also	better → bet ❌	better → good ✅
Linguistic awareness	No (rule-based chopping)	Yes (uses vocabulary + grammar)
Accuracy	Lower	Higher
Speed	Faster	Slower
Output	May not be a real word	Always a valid word
Use case	Quick text normalization, search indexing	NLP tasks needing meaning (chatbots, sentiment analysis, ML models)

Example

Sentence:
“The children are running better today.”

Stemming:
children → children
running → run
better → bet ❌

Lemmatization:
children → child
running → run
better → good ✅

In short:

Use stemming when speed matters more than precision.

Use lemmatization when understanding meaning matters.

```python
stemming = PorterStemmer()

for word in words:
    print(word + " --> " + stemming.stem(word))
```

```python
stemming.stem("congratulations")
```

```python
stemming.stem("sitting")
```

```python
## Regular Expression Stemmer
from nltk.stem import RegexpStemmer
```

```python
regex_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
regex_stemmer.stem('eating')
```

```python
regex_stemmer.stem('ingeating')
```

```python
## snowball stemmer 
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

for word in words:
    print(word + " --> " + stemmer.stem(word))
```
